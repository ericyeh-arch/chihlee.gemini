## python數學運算

```python
# 使用雙引號包住的字串 (string)
print("這是我的第1個函式") #str
# 使用單引號包住的字串 (string)
print('這是我的第2個函式') # str
# 整數 (integer)
print(50) #int
# 浮點數 (float)
print(50.0) #float
```

```python
# print() 可同時輸出多個值，之間以逗號分隔
# 輸出文字與整數
print("這是整數", 50)
# 輸出文字與浮點數
print("這是浮點數", 50.5)
# 輸出文字與字串
print("這是字串", 'robert')
# 輸出布林值 True (真)
print("這是布林值", True)
# 輸出布林值 False (假)
print("這是布林值", False)
```

```python
# 加法運算
print(3 + 5)
# 多個數字相加
print(3 + 5 + 10)
# 先乘除後加減：先計算 5 * 2 = 10，再加 3
print( 3 + 5 * 2)
# 括號優先：先計算 3 + 5 = 8，再乘 2
print((3 + 5) * 2)
# 指數運算：99 的 99 次方
print(99 ** 99)
```

```python
# 匯入數學模組，可取得圓周率等數學常數與函式
import math

# 使用 input() 讓使用者輸入半徑，float() 將輸入的文字轉成浮點數
radius = float(input("請輸入半徑:"))
# 計算圓面積 = 半徑的平方 × 圓周率
area = radius ** 2 * math.pi
# 輸出計算結果
print("半徑:", radius, ", 圓面積=", area)
```
